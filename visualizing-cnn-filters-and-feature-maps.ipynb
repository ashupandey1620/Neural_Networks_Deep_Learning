```python
from keras.applications.vgg16 import VGG16
#Load the model
model = VGG16()
```

```python
model.summary()
```

```python
from keras.utils import plot_model

plot_model(model)
```

```python
for i in range(len(model.layers)):
	# check for convolutional layer
	if 'conv' not in model.layers[i].name:
		continue
	# get filter weights
	filters, biases = model.layers[i].get_weights()
	print("layer number",i,model.layers[i].name, filters.shape)
```

```python
# retrieve weights from the second hidden layer
filters , bias = model.layers[1].get_weights()
```

```python
# normalize filter values to 0-1 so we can visualize them
f_min, f_max = filters.min(), filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as plt
```

```python
n_filters = 6
ix = 1

fig = plt.figure(figsize=(15,10))

for i in range(n_filters):

    # get the filters
    f = filters[:, :, :, i]

    # plot each channel separately
    for j in range(3):

        ax = fig.add_subplot(n_filters, 3, ix)

        ax.set_xticks([])
        ax.set_yticks([])

        plt.imshow(f[:, :, j], cmap='gray')

        ix += 1

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import (
    load_img,
    img_to_array
)

# -------------------------------
# 1. Check model layers
# -------------------------------

for layer in model.layers:
    print(layer.name)

# -------------------------------
# 2. Create visualization model
# -------------------------------

visualization_model = Model(
    inputs=model.inputs,
    outputs=model.get_layer('block1_conv1').output
)

# -------------------------------
# 3. Load image
# -------------------------------

image_path = "/kaggle/input/datasets/ashupandey1620/kohli-pic/kohli.jpg"

image = load_img(
    image_path,
    target_size=(224, 224)
)

# Display original image
plt.figure(figsize=(5,5))
plt.imshow(image)
plt.axis('off')
plt.title("Original Image")
plt.show()

# -------------------------------
# 4. Preprocess image
# -------------------------------

image = img_to_array(image)

# Normalize
image = image / 255.0

# Expand dimensions
image = np.expand_dims(image, axis=0)

print("Input Shape:", image.shape)

# -------------------------------
# 5. Generate feature maps
# -------------------------------

features = visualization_model.predict(image)

print("Feature Map Shape:", features.shape)

# -------------------------------
# 6. Visualize feature maps
# -------------------------------

fig = plt.figure(figsize=(20,15))

n_features = min(features.shape[3], 64)

for i in range(n_features):

    plt.subplot(8, 8, i + 1)

    plt.imshow(features[0, :, :, i], cmap='gray')

    plt.axis('off')

plt.tight_layout()
plt.show()
```
